# Two Pointer and Sliding Window

## 1423. Maximum Points You Can Obtain from Cards

```python
class MaxPoints:
    def maxScore(self, arr, k: int) -> int:
        l = k-1
        r = len(arr)-1
        lsum = sum(arr[:k]) #Adding up the first k elements
        rsum = 0
        maxsum = lsum
        for i in range(k-1, -1, -1): #Starting from 4, going until -1(till 0), reducing one step at a time(i--)
            lsum = lsum - arr[i] #Subtracting one index value as we are sliding the window to left.
            rsum = rsum + arr[r] #Adding at the same time on the right side by one index value at a time.
            l = l-1
            r = r-1
        maxsum = max(maxsum , (lsum+rsum))
        return maxsum

sol = MaxPoints()
sol.maxScore([1,2,3,4,5,6,1], 3)
```

## 3. Longest Substring Without Repeating Characters

```python
class LongestSub:
    def lengthOfLongestSubstring(self, s: str) -> int:
        l = 0
        r = 0
        maxstr = ''
        curmaxstr = ''
        while r<len(s):
            if s[r] not in curmaxstr:
                curmaxstr = curmaxstr + s[r]
                r = r+1
                if len(curmaxstr) > len(maxstr):
                    maxstr = curmaxstr
                    continue
            else:
                curmaxstr = curmaxstr[1:] # Instead of checking all possible combinations, just slide to the next window when you find a duplicate.
                # This way you'll reach to the sub string quick before iterating all sub strings starting from left index that are less than the length of the existing maxstr.
                l = l+1

        print(maxstr)
        return len(maxstr)

sol = LongestSub()
sol.lengthOfLongestSubstring("abcabcbb")
sol.lengthOfLongestSubstring("bbbbbbbb")
sol.lengthOfLongestSubstring("pwwkew")
```

## 1004. Maximum Consecutive Ones III

```python
class Solution:
    def longestOnes(self, nums: List[int], k: int) -> int:
        l=0
        r=0
        curmax=0
        fmax=0
        flip_count = 0
        for r in range(0,len(nums)):
            if nums[r] == 0:
                flip_count = flip_count + 1

            while flip_count > k:
                if nums[l] == 0:
                    flip_count = flip_count -1
                l = l+1
            fmax = max(fmax, r-l+1)
        return fmax



```
