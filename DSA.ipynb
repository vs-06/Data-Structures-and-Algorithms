## 1423. Maximum Points You Can Obtain from Cards

```python
class MaxPoints:
    def maxScore(self, arr, k: int) -> int:
        l = k-1
        r = len(arr)-1
        lsum = sum(arr[:k]) #Adding up the first k elements
        rsum = 0
        maxsum = lsum
        for i in range(k-1, -1, -1): #Starting from 4, going until -1(till 0), reducing one step at a time(i--)
            lsum = lsum - arr[i] #Subtracting one index value as we are sliding the window to left.
            rsum = rsum + arr[r] #Adding at the same time on the right side by one index value at a time.
            l = l-1
            r = r-1
        maxsum = max(maxsum , (lsum+rsum))
        return maxsum

sol = MaxPoints()
sol.maxScore([1,2,3,4,5,6,1], 3)
```

## 3. Longest Substring Without Repeating Characters

```python
class LongestSub:
    def lengthOfLongestSubstring(self, s: str) -> int:
        l = 0
        r = 0
        maxstr = ''
        curmaxstr = ''
        while r<len(s):
            if s[r] not in curmaxstr:
                curmaxstr = curmaxstr + s[r]
                r = r+1
                if len(curmaxstr) > len(maxstr):
                    maxstr = curmaxstr
                    continue
            else:
                curmaxstr = curmaxstr[1:] # Instead of checking all possible combinations, just slide to the next window when you find a duplicate.
                # This way you'll reach to the sub string quick before iterating all sub strings starting from left index that are less than the length of the existing maxstr.
                l = l+1

        print(maxstr)
        return len(maxstr)

sol = LongestSub()
sol.lengthOfLongestSubstring("abcabcbb")
sol.lengthOfLongestSubstring("bbbbbbbb")
sol.lengthOfLongestSubstring("pwwkew")
```

## 1004. Maximum Consecutive Ones III

```python
class Solution:
    def longestOnes(self, nums: List[int], k: int) -> int:
        l=0
        r=0
        curmax=0
        fmax=0
        flip_count = 0
        for r in range(0,len(nums)):
            if nums[r] == 0:
                flip_count = flip_count + 1

            while flip_count > k:
                if nums[l] == 0:
                    flip_count = flip_count -1
                l = l+1
            fmax = max(fmax, r-l+1)
        return fmax



```

## 904. Fruit Into Baskets

```python
'''
THE LOGIC :
Brute force checks every possible subarray and counts how many distinct fruit types it contains, then returns the longest one with at most two types. This works, but it’s inefficient because the number of subarrays is large.

The optimized approach uses a sliding window and a dictionary (`picked_types`) to track fruit counts in the current window. As we move `r` from left to right across the array:

* If `fruits[r]` already exists in `picked_types`, increment its count; otherwise add it with count `1`.
* After adding each fruit, check `len(picked_types)`.

  * If it’s `<= 2`, the window is valid (we can pick from these trees with two baskets).
  * If it becomes `3`, the window is invalid because we now have three fruit types.

To restore validity, we shrink the window from the left:

* While `len(picked_types) > 2`, decrement the count for `fruits[l]` and move `l` right by 1.
* If any fruit’s count drops to `0`, remove that fruit type from the dictionary entirely. This represents fully “dropping” one fruit type from the window.

After each step (once the window is valid), update the answer with the current window length `r - l + 1`.

Why this is efficient:

* Each index moves into the window once (via `r`) and out of the window once (via `l`), so the total work is linear.
* Time complexity: **O(n)**
* Space complexity: **O(1)** in practice, because the dictionary holds at most 2 fruit types at any valid moment (temporarily 3 during adjustment).


'''
from typing import List

class Solution:
    def totalFruit(self, fruits: List[int]) -> int:
        l = 0
        r = 0
        picked_types = {}
        final_max = 0

        while r < len(fruits):
            if fruits[r] in picked_types:
                picked_types[fruits[r]] = picked_types[fruits[r]] + 1
            else:
                picked_types[fruits[r]] = 1
            r = r + 1

            while len(picked_types) > 2:
                left_fruit = fruits[l]
                picked_types[left_fruit] = picked_types[left_fruit] - 1
                if picked_types[left_fruit] == 0:
                    del picked_types[left_fruit]
                l = l + 1

            if r - l > final_max:
                final_max = r - l

        return final_max
```

## Implementing Stack and Queue in Python

```python
import numpy as np

class Stack:
  def __init__(self, capacity: int = 5):
    # Initialize instance variables within the constructor
    self.stack_array = np.zeros(capacity, dtype=int)
    self.top_index = -1
    self.capacity = capacity

  def push(self, num: int) -> list:
    if self.top_index < self.capacity - 1:
      self.top_index += 1
      self.stack_array[self.top_index] = num
      return self.stack_array[:self.top_index+1].tolist()
    else:
      print('Error: Cannot add element, the stack is full.')
      return self.stack_array[:self.top_index+1].tolist()

  def pop(self) -> list:
    if self.top_index > -1:
      # If you intended `pop` to return the value itself, it would be `return self.stack_array[self.top_index]`
      self.stack_array[self.top_index] = 0
      self.top_index -= 1
      return self.stack_array[:self.top_index+1].tolist() # Return current stack state
    else:
      print('Error: Cannot remove element, the stack is empty.')
      return []

  def top(self) -> int:
    if self.top_index > -1:
      return int(self.stack_array[self.top_index])
    else:
      print('Error: Stack is empty.')
      return None

  def is_empty(self) -> bool:
    return self.top_index == -1

  def size(self) -> int:
    return self.top_index + 1

# Example usage:
my_stack = Stack(capacity=3)
print(f"Initial stack (empty): {my_stack.stack_array[:my_stack.top_index+1].tolist()}")

print(f"Push 10: {my_stack.push(10)}")
print(f"Push 20: {my_stack.push(20)}")
print(f"Top element: {my_stack.top()}")
print(f"Push 30: {my_stack.push(30)}")
print(f"Push 40: {my_stack.push(40)}") # This should print an error and return the full stack

print(f"Pop: {my_stack.pop()}")
print(f"Top element: {my_stack.top()}")
print(f"Pop: {my_stack.pop()}")
print(f"Pop: {my_stack.pop()}")
print(f"Pop: {my_stack.pop()}") # This should print an error and return an empty list
print(f"Is stack empty? {my_stack.is_empty()}")
print(f"Top element: {my_stack.top()}") # This should print an error and return None
```

```python
import numpy as np
class Queue:
  def __init__(self, capacity:int = 5) -> None:
    self.queue_array = np.zeros(capacity, dtype=int)
    self.start = -1
    self.end = -1
  def push(self, num:int) -> list:
    if self.end <= self.capacity:
      self.end = (self.end + 1)% self.capacity
      self.queue_array[self.end] = num
      return self.queue_array[:self.end+1].tolist()
    else:
      print('Error: Cannot add element, the queue is full.')
      return []
  def pop(self) -> list:
    if self.start < self.end:
      self.start = self.start + 1





```

## Sorting Algorithms implementation.

```python
# Bubble sort
 ## Time: O(n2), Space: O(1)

def bubble_sort(arr):
  n = len(arr)

  for i in range(n):
    swapped = False

    for j in range(0, n-i-1):
      if arr[j] > arr[j+1]:
        arr[j] , arr[j+1] = arr[j+1] , arr[j]
        swapped = True

    if not swapped:
      break
  return arr

bubble_sort([5,4,26,1,9,84,41])
```
